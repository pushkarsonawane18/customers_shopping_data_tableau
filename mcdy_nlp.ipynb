# Text Preprocessing

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
print(stopwords.words('english'))
```

```python
stop_words = set(stopwords.words('english'))
```

```python
import string
```

```python
def text_preprocess(text):
  text = text.lower() # for lowercase
  text = text.translate(str.maketrans('','',string.punctuation)) # for punctuations
  tokens = word_tokenize(text) # for tokenize
  tokens = [t for t in tokens if t not in stop_words]
  return ' '.join(tokens)
```

```python
df['cleaned_review'] = df['review'].dropna().apply(text_preprocess)
```

```python
from nltk.stem import PorterStemmer # stemming
from nltk.stem import WordNetLemmatizer # lemmitization
```

```python
stemmer = PorterStemmer()
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = ['running','walking','cooking','happily','studying','reading','writing','sleeping','screaming','screening','streaming','dogs','cats']
```

```python
print("word   stemming    lemmitization")
print("-")
for word in words:
  print(f"{word:<11} {stemmer.stem(word):<11}  {lemmatizer.lemmatize(word)}")
```
